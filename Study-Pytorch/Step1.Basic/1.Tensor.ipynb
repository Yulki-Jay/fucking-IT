```python
import torch
import numpy as np
```

```python
data = [[1,2],[3,4]]
x_data = torch.tensor(data)
x_data
```

```python
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
x_np
```

```python
x_ones = torch.ones_like(x_data)
print(f"Ones Tensor: \n {x_ones} \n")

x_rand = torch.rand_like(x_data, dtype=torch.float)
print(f"Random Tensor: \n {x_rand} \n")
```

```python
shape = (2,3,)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor} \n")
```

```python
tensor = torch.rand(3,4)
print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```

```python
if torch.accelerator.is_available():
    tensor = tensor.to(torch.accelerator.current_accelerator())
print(f"Device tensor is stored on: {tensor.device}")
```

```python
tensor = torch.randn(4,4)
print(f"Original tensor: \n {tensor} \n")
print(f"First row: {tensor[0]}")
print(f"First column: {tensor[:,0]}")
print(f"Last column: {tensor[:,-1]}")
tensor[:,1] = 0
print(f"Modified tensor: \n {tensor}")

```

```python
t1 = torch.cat([tensor, tensor,tensor], dim=0)
print(t1)
print(t1.shape)
```

```python
tensor = torch.rand(3,4).to(torch.accelerator.current_accelerator() if torch.accelerator.is_available() else 'cpu')
print(f"Shape of tensor: {tensor.shape}")
print(f"Tensor: {tensor}")
```

```python
y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)

y3 = torch.rand_like(tensor)
print(f"y1: {y1} \n y2: {y2} \n y3: {y3}")
torch.matmul(tensor, tensor.T,out=y3)
print(f"y3: {y3}")

z1 = tensor * tensor
z2 = tensor.mul(tensor)
z3 = torch.rand_like(tensor)
print(f"z1: {z1} \n z2: {z2} \n z3: {z3}")
torch.mul(tensor, tensor, out=z3)
print(f"z3: {z3}")
```

```python
agg = tensor.sum()
print(agg)
print(agg.item(),type(agg),type(agg.item()))  # get the Python number from a tensor containing a single value
```

```python
print(tensor)
tensor.add_(5)
print(tensor)
```

```python
t = torch.ones(5)
print(f"t: {t}, type: {type(t)}")

n = t.numpy()
print(f"n: {n}, type: {type(n)}")
```

```python
# 这是因为它们共享内存空间，对一个进行修改会影响另一个
t.add_(5)
print(f"t: {t}, type: {type(t)}")
print(f"n: {n}, type: {type(n)}")
```

```python
n += 5
print(f"t: {t}, type: {type(t)}")
print(f"n: {n}, type: {type(n)}")
```
